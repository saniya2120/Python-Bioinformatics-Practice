```python
probe_ids = data["ID_REF"]

expression_values = data.drop(columns=["ID_REF"])

print("Probe IDs:", probe_ids.shape)
print("Expression data:", expression_values.shape)
```

```python
sample_ids = expression_values.columns.tolist()

print("Total samples:", len(sample_ids))
print(sample_ids)
```

```python
sample_metadata = {
    "Control": [f"GSM{n}" for n in range(577963, 577972)],
    "Type 2 Diabetes": [f"GSM{n}" for n in range(577990, 577998)]
}

print("Control samples:", sample_metadata["Control"])
print("Type 2 Diabetes samples:", sample_metadata["Type 2 Diabetes"])
```

```python
sample_groups = []

for sample in sample_ids:
    if sample in sample_metadata["Control"]:
        sample_groups.append("Control")
    elif sample in sample_metadata["Type 2 Diabetes"]:
        sample_groups.append("Type 2 Diabetes")
    else:
        sample_groups.append("Other")

metadata = pd.DataFrame({
    "Sample_ID": sample_ids,
    "Condition": sample_groups
})

metadata
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.boxplot(expression_values.values, tick_labels=expression_values.columns)

plt.xticks(rotation=90)
plt.ylabel("Expression value")
plt.title("Expression Distribution Across Samples")

plt.savefig("Day5/type-2-diabetes-gene-expression-analysis/figures/expression_distribution_boxplot.png")

plt.show()
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Keep only Control and Type 2 Diabetes samples
selected_samples = metadata[
    metadata["Condition"].isin(["Control", "Type 2 Diabetes"])
]["Sample_ID"].tolist()

pca_data = expression_values[selected_samples].T

# Standardize the data
scaled_data = StandardScaler().fit_transform(pca_data)

# Perform PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

print("PCA shape:", pca_result.shape)
print("Explained variance:", pca.explained_variance_ratio_)
```

```python
import matplotlib.pyplot as plt

# Create labels for each sample
conditions = metadata[
    metadata["Condition"].isin(["Control", "Type 2 Diabetes"])
]["Condition"].tolist()

# Plot Control samples
for condition, marker in [("Control", "o"), ("Type 2 Diabetes", "^")]:
    indices = [i for i, c in enumerate(conditions) if c == condition]

    plt.scatter(
        pca_result[indices, 0],
        pca_result[indices, 1],
        label=condition,
        marker=marker,
        s=80
    )

# Label each sample
for i, sample in enumerate(selected_samples):
    plt.text(
        pca_result[i, 0],
        pca_result[i, 1],
        sample.replace("GSM", ""),
        fontsize=8
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA of Control and Type 2 Diabetes Samples")
plt.legend()
plt.show()
```
